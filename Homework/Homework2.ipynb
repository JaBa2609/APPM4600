```python
import numpy as np
import matplotlib.pyplot as plt

def fixed_form(x,delta):
    return -2*np.sin(x+delta/2)*np.sin(delta/2)

def original_form(x,delta):
    return np.cos(x+delta)-np.cos(x)

def custom_algo(x,delta):
    return -delta*np.sin(x) - (delta**2/2)*np.cos(x)

def custom_algo2(x,delta):
    return delta*np.cos(x) - (delta**2/2)*np.sin(x)

delta_list = [10**(x) for x in range(-16,1)]
delta_list

x1 = np.pi
x2 = 10**(-6)

def diff_plot(x,del_list):
    fixed_outputs = [fixed_form(x, d) for d in del_list]
    original_outputs = [original_form(x, d) for d in del_list]
    custom_outputs = [custom_algo(x, d) for d in del_list]
    custom_outputs2 = [custom_algo2(x, d) for d in del_list]
    
    diff_outputs = [fixed_outputs[i] - original_outputs[i] for i in range(len(fixed_outputs))]
    diff_outputs2 = [custom_outputs[i] - original_outputs[i] for i in range(len(original_outputs))]
    diff_outputs3 = [custom_outputs2[i] - original_outputs[i] for i in range(len(original_outputs))]

    
    plt.plot(del_list, diff_outputs, label=f'fixed - original = {x}')
    plt.plot(del_list, diff_outputs2, label=f'custom1 - original = {x}')
    #plt.plot(del_list, diff_outputs3, label=f'custom2 - original = {x}')
    plt.xscale('log')
    plt.xlabel('delta')
    plt.title('fixed_form - original_form output')

#plot the difference of fixed-original trig expressions for x_1 and x_2
diff_plot(x1,delta_list)
diff_plot(x2,delta_list)
plt.legend()
plt.show()
```
