```python
import numpy as np
import matplotlib.pyplot as plt

def p_9(x):
    return x**9 - 18*x**8 + 144*x**7 -672*x**6 + 2016*x**5-4032*x**4+5376*x**3-4608*x**2 + 2304*x - 512
    
x_vals = np.arange(1.920,2.080,.001)
#print(x_vals)
#print(len(x_vals))
y_vals = np.array([])
for i in range(len(x_vals)):
    #print(i)
    #print(p_9(x_vals[i]))
    y_vals = np.append(y_vals, p_9(x_vals[i]))


def closed_form(x):
    return (x-2)**9
    
x_vals2 = np.arange(1.920,2.080,.001)
y_vals2 = np.array([])

for i in range(len(x_vals2)):
    y_vals2 = np.append(y_vals2, closed_form(x_vals2[i]))
    
plt.plot(x_vals2,y_vals2)
plt.plot(x_vals,y_vals)
plt.show()
```
