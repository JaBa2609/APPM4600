```python
#prototype program to make matrix multiplication

matrix1 = np.array([[1,0], 
                  [1,2]])

matrix2 = np.array([[1,2],
                    [3,4]])

# need to do: rows of first times all columns of second
def matrixMult(x,y):
#   Computes the dot product of the n x 1 vectors x and y

    size1 = x.shape
    size2 = y.shape
    if size1[1] != size2[1]:
        print('Error on inner dimension sizes')
        return

    #print(size1,size2)

    result = np.zeros((2,2))
    #print(result)

    for i in range(size1[0]):
        for j in range(size2[0]):
            result[i,j] = dotProduct(x[i,:], y[:,j],size1[0])

    return result

matrixMult(matrix1,matrix2)
```
