```python
import numpy as np

def order_of_convergence(p_true, p_hat):
    """
    Numerically estimate the order of convergence alpha of an iterative
    method, given the true fixed point p_true and a sequence of
    approximations p_hat = [p0, p1, ..., pN].

    If the method converges with order alpha:
        e_{n+1} ~ C * e_n^alpha        where e_n = |p_hat[n] - p_true|

    Taking logs of three consecutive errors:
        alpha ~ ln(e_{n+1}/e_n) / ln(e_n/e_{n-1})
    """
    p_hat = np.asarray(p_hat, dtype=float)
    errors = np.abs(p_hat - p_true)

    alphas = []
    for n in range(1, len(errors) - 1):
        e_prev, e_curr, e_next = errors[n - 1], errors[n], errors[n + 1]
        if e_prev == 0 or e_curr == 0 or e_next == 0:
            alphas.append(np.nan)
            continue
        alpha = np.log(e_next / e_curr) / np.log(e_curr / e_prev)
        alphas.append(alpha)

    return np.array(alphas), errors
```

```python
import numpy as np

# ---------------------------------------------------------------
# Part (a): Fixed point iteration
# ---------------------------------------------------------------
def g(x):
    return (10 / (x + 4)) ** 0.5

def fixed_point_iteration(p0, tol=1e-10, max_iter=1000):
    """Returns (converged_value, n_iterations, full_history_array)."""
    p_old = p0
    history = [p0]
    for i in range(1, max_iter + 1):
        p_new = g(p_old)
        history.append(p_new)
        if abs(p_new - p_old) < tol:
            return p_new, i, np.array(history)
        p_old = p_new
    return p_old, max_iter, np.array(history)


# ---------------------------------------------------------------
# Part (b): Order of convergence
# ---------------------------------------------------------------
def order_of_convergence(p_true, p_hat):
    p_hat = np.asarray(p_hat, dtype=float)
    errors = np.abs(p_hat - p_true)
    alphas = []
    for n in range(1, len(errors) - 1):
        e_prev, e_curr, e_next = errors[n - 1], errors[n], errors[n + 1]
        if e_prev == 0 or e_curr == 0 or e_next == 0:
            alphas.append(np.nan)
            continue
        alpha = np.log(e_next / e_curr) / np.log(e_curr / e_prev)
        alphas.append(alpha)
    return np.array(alphas), errors


# ---------------------------------------------------------------
# Run both together
# ---------------------------------------------------------------
p_true = 1.3652300134140976
p0 = 1.5
tol = 1e-10

root, n_iter, p_hat = fixed_point_iteration(p0, tol=tol)   # (a) generates the sequence
alphas, errors = order_of_convergence(p_true, p_hat)        # (b) consumes it directly

print(f"Converged value: {root}, in {n_iter} iterations")
print(f"Estimated order of convergence: {np.mean(alphas[1:-1]):.4f}")
```
